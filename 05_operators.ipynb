```python
text = "Data Science"
vowels = "aeiouAEIOU"
count = 0
for ch in text:
    if ch in vowels:
        count += 1
print("Number of vowels =", count)
```

Build a simple calculator using arithmetic and comparison operators.

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
op = input("Enter operator (+, -, *, /): ")
if op == "+":
    print("Result =", a + b)
elif op == "-":
    print("Result =", a - b)
elif op == "*":
    print("Result =", a * b)
elif op == "/":
    if b != 0:
        print("Result =", a / b)
    else:
        print("Division by zero is not possible")
else:
    print("Invalid operator")
```

Create a grading system using comparison and logical operators.

 Build a menu-driven calculator that uses arithmetic, comparison, logical, assignment, membership, identity, and bitwise operators wherever applicable.

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print("1.Addition")
print("2.Subtraction")
print("3.Multiplication")
print("4.Division")
print("5.Bitwise AND")
print("6.Check Equal")
print("7.Check Identity")
print("8.Check Membership")
choice = int(input("Enter your choice: "))
if choice == 1:
    result = a
    result += b                      # Assignment operator
    print("Addition =", result)
elif choice == 2:
    print("Subtraction =", a - b)
elif choice == 3:
    print("Multiplication =", a * b)
elif choice == 4:
    if b != 0:                           # Comparison operator
        print("Division =", a / b)
    else:
        print("Division not possible")
elif choice == 5:
    print("Bitwise AND =", a & b)         # Bitwise operator
elif choice == 6:
    if a == b:                            # Comparison operator
        print("Numbers are Equal")
    else:
        print("Numbers are Not Equal")
elif choice == 7:
    print(a is b)                          # Identity operator
elif choice == 8:
    numbers = [10, 20, 30, 40, 50]
    if a in numbers:                        # Membership operator
        print("First number is in the list")
    else:
        print("First number is not in the list")
else:
    print("Invalid Choice")
```
