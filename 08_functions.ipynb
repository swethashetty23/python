Assignment 35: Fibonacci Function
Write a function named generate_fibonacci().

Requirements
Accept the number of terms.
Return the complete Fibonacci series as a list.
Handle zero and negative term counts.
Do not use recursion in this assignment.

```python
def generate_fibonacci(terms):
  if terms <= 0:
    return []
  elif terms == 1:
    return [0]
  fibonacci = [0, 1]
  first = 0
  second = 1
  for i in range(terms-2):
    next_number = first + second
    fibonacci.append(next_number)
    first = second
    second = next_number
  return fibonacci
n = int(input("enter number of terms: "))
res = generate_fibonacci(n)
res
```

```python
employees = [

{
    "id": 101,
    "name": "Ravi",
    "department": "IT",
    "salary": 60000,
    "performance_score": 88,
    "experience": 5
},

{
    "id": 102,
    "name": "Anitha",
    "department": "HR",
    "salary": 45000,
    "performance_score": 72,
    "experience": 4
},

{
    "id": 103,
    "name": "Kiran",
    "department": "IT",
    "salary": 75000,
    "performance_score": 94,
    "experience": 7
},

{
    "id": 104,
    "name": "Meena",
    "department": "Finance",
    "salary": 55000,
    "performance_score": 81,
    "experience": 6
},

{
    "id": 105,
    "name": "Arjun",
    "department": "IT",
    "salary": 40000,
    "performance_score": 65,
    "experience": 2
}
]
def display_employees(emp_list):
  if not emp_list:
    print("Employee list is empty:")
    return
    print("Employees details:")
  for  emp in emp_list:
      print("\nID:", emp["id"])
      print("Name:", emp["name"])
      print("Department:", emp["department"])
      print("Salary:", emp["salary"])
      print("performance_score:", emp["performance_score"])
      print("Experience:", emp["experience"])
display_employees(employees)
```

```python
#part2 for dipalying employee names
def display_employee_names(emp_list):
  names = list(map(lambda emp: emp["name"], emp_list))
  for name in names:
    print(name)
display_employee_names(employees)
```

```python
def display_high_performers(emp_list):
  high_performers = list(filter(lambda emp: emp["performance_score"] >= 80, emp_list))
  print("\nHigh performers:")
  for emp in high_performers:
    print("Name:", emp["name"])
    print("Performance Score:", emp["performance_score"])
display_high_performers(employees)
```

```python
#part4
def search_dept(emp_list, department):
  dept_emp = list(filter(lambda emp: emp["department"].lower() == department.lower(), emp_list))
  print("\n Employees in department:", department)
  if dept_emp:
    for emp in dept_emp:
      print("\nID:", emp["id"])
      print("Name:", emp["name"])
      print("Salary:", emp["salary"])
      print("Performance score:", emp["performance_score"])
      print("Experience:", emp["experience"])
  else:
    print("department not found")
department = input("enter department name: ")
search_dept(employees, department)
```

```python
# part5
def calculate_revised_salaries(emp_list):
  def revise_salary(emp):
    update_emp = emp.copy()
    score = emp["performance_score"]
    if score > 90:
      update_emp["salary"] = emp["salary"] * 1.15
    elif score > 80:
      update_emp["salary"] = emp["salary"]  * 1.10
    elif score > 70:
      update_emp["salary"] = emp["salary"] * 1.05
    else:
      update_emp["salary"] = emp["salary"]
    return update_emp
  revised_employees = list(map(revise_salary, emp_list))
  for emp in revised_employees:
    print("\nName: ", emp["name"])
    print("Salary:", emp_list[revised_employees.index(emp)]["salary"])
    print("Updated Salary:", emp["salary"])
  return revised_employees
calculate_revised_salaries(employees)
```

```python

from functools import reduce
def calculate_total_payroll(emp_list, revised_list):
  current_total_payroll = reduce(lambda total, emp: total + emp["salary"], emp_list, 0)
  revised_total_payroll = reduce(lambda total, emp: total + emp["salary"], revised_list, 0)
  additional_expense = revised_total_payroll - current_total_payroll
  print("Current salary:", current_total_payroll)
  print("Revised salary:", revised_total_payroll)
  print("Additional Expenses:", additional_expense)
revised_employees= calculate_revised_salaries(employees)
calculate_total_payroll(employees, revised_employees)
```

```python
def highest_paid_emp(emp_list):
  highest = reduce(lambda emp1, emp2: emp1 if emp1["salary"] > emp2["salary"] else emp2, emp_list)
  print("Employee: ")
  print("ID:", highest["id"])
  print("Name:", highest["name"])
  print("Department:", highest["department"])
  print("Salary:", highest["salary"])
  print("Performance Score:", highest["performance_score"])
  print("Experience:", highest["experience"])
highest_paid_emp(employees)
```

```python
def sort_emp(emp_list):
  name = sorted(emp_list, key = lambda emp: emp["name"])
  for emp in name:
    print(emp["name"])
  print("\nSorted by salary:")
  salary = sorted(emp_list, key = lambda emp: emp["salary"])
  for emp in salary:
    print(emp["name"], "-", emp["salary"])

  print("\nSorted by performance: ")
  performance = sorted(emp_list, key = lambda emp: emp["performance_score"])
  for emp in performance:
    print(emp["name"], "-", emp["performance_score"])


  print("\nSorted by experience: ")
  experience = sorted(emp_list, key = lambda emp: emp["experience"])
  for emp in experience:
    print(emp["name"], "-", emp["experience"])


sort_emp(employees)
```

```python
def bonus_calculator(bonus_percentage):
  def calculate_bonus(salary):
    return salary * bonus_percentage / 100
  return calculate_bonus
bonus_percent_5 = bonus_calculator(5)
bonus_percent_10 = bonus_calculator(10)
bonus_percent_15 = bonus_calculator(15)
print("5% bonus:", bonus_percent_5(10000))
print("10% bonus:", bonus_percent_10(10000))
print("15% bonus:", bonus_percent_15(10000))
```

```python
def generate_department_report(emp_list, department):
    dept_emp = list(filter(lambda emp: emp["department"].lower() == department.lower(), emp_list))
    if not dept_emp:
        return {"Department not found"}
    def emp_count():
        return len(dept_emp)
    def total_salary():
        return sum(emp["salary"] for emp in dept_emp)
    def avg_salary():
        return total_salary() / emp_count()
    def avg_performance():
        total_score = sum(emp["performance_score"] for emp in dept_emp)
        return total_score / emp_count()
    report = {
        "Department": department,
        "Number of Employees": emp_count(),
        "Total Salary": total_salary(),
        "Average Salary": avg_salary(),
        "Average Performance Score": avg_performance()
    }
    return report
department = input("Enter Department: ")
report = generate_department_report(employees, department)
for key, value in report.items():
    print(key, ":", value)
```

```python
company_name = "Edukron"         # global
def generate_department_report(emp_list, department):
    dept_name = department #enclosing
    dept_emp = list(filter(lambda emp: emp["department"].lower() == dept_name.lower(), emp_list))
    if not dept_emp:
        return { "error":"Department not found"}
    def create_report():
      report_title = "Department Report"
      return {
            "Company": company_name,          # Global variable
            "Title": report_title,            # Local variable
            "Department": dept_name,    # Enclosing variable
            "Employees": len(dept_emp),
            "Total Salary": sum(emp["salary"] for emp in dept_emp),
            "Avg salary": round(sum(emp["salary"] for emp in dept_emp) / len(dept_emp), 2 ),
            "Avg Performance": round(sum(emp["performance_score"] for emp in dept_emp) / len(dept_emp), 2)
       }
    return create_report()
department = input("Enter Department: ")
report = generate_department_report(employees, department)
for key, value in report.items():
    print(key, ":", value)
```
