Create a scientific calculator that supports square root, factorial, logarithm, power, trigonometric functions, GCD, and LCM.


```python
print("1. Square Root")
print("2. Factorial")
print("3. Log")
print("4. Power")
print("5. GCD")
print("6. LCM")
choice = int(input("Enter choice: "))
if choice == 1:
    n = float(input("Enter number: "))
    print(math.sqrt(n))
elif choice == 2:
    n = int(input("Enter number: "))
    print(math.factorial(n))
elif choice == 3:
    n = float(input("Enter number: "))
    print(math.log(n))
elif choice == 4:
    a = float(input("Base: "))
    b = float(input("Exponent: "))
    print(math.pow(a, b))
elif choice == 5:
    a = int(input("First number: "))
    b = int(input("Second number: "))
    print(math.gcd(a, b))
elif choice == 6:
    a = int(input("First number: "))
    b = int(input("Second number: "))
    print(abs(a * b) // math.gcd(a, b))
else:
    print("Invalid choice")
```
